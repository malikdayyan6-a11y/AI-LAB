```python
MAX, MIN = float("inf"), float("-inf")

tree = {
    "A": [("B", "min"), ("C", "min")],
    "B": [("D", "max"), ("E", "max")],
    "C": [("F", "max"), ("G", "max")],
    "D": [("H", "leaf", 4), ("I", "leaf", 2)],
    "E": [("J", "leaf", -3), ("K", "leaf", -6)],
    "F": [("L", "leaf", 7), ("M", "leaf", 0)],
    "G": [("N", "leaf", 5), ("O", "leaf", 8)]
}

def alphabeta(node, node_type, alpha, beta):
    if node_type == "leaf":
        return node
    
    children = tree[node]
    
    if node_type == "max":
        value = MIN
        for child in children:
            cname = child[0]
            ctype = child[1]
            
            if ctype == "leaf":
                child_value = child[2]
            else:
                child_value = alphabeta(cname, ctype, alpha, beta)
            
            value = max(value, child_value)
            alpha = max(alpha, value)
            print(f"visited {cname} with value {child_value}, alpha={alpha}, beta={beta}")
            
            if beta <= alpha:
                print(f"Pruned remaining children of {node} (max node)")
                break
        
        return value
    
    elif node_type == "min":
        value = MAX
        for child in children:
            cname = child[0]
            ctype = child[1]
            
            if ctype == "leaf":
                child_value = child[2]
            else:
                child_value = alphabeta(cname, ctype, alpha, beta)
            
            value = min(value, child_value)
            beta = min(beta, value)
            print(f"visited {cname} with value {child_value}, alpha={alpha}, beta={beta}")
            
            if beta <= alpha:
                print(f"Pruned remaining children of {node} (min node)")
                break
        
        return value

root_value = alphabeta("A", "max", MIN, MAX)
print("Optimal value at root A:", root_value)
```

```python
import math

AI = "X"
HUMAN = "O"
EMPTY = ""

board = [EMPTY] * 9

def print_board():
    for r in range(3):
        row = board[3*r:3*r+3]
        print(" | ".join(row))
        if r < 2:
            print("-" * 9)
    print()

def winner(b):
    wins = [
        (0,1,2), (3,4,5), (6,7,8),
        (0,3,6), (1,4,7), (2,5,8),
        (0,4,8), (2,4,6)
    ]
    for a,b,c in wins:
        if board[a] == board[b] == board[c] != EMPTY:
            return board[a]
    if EMPTY not in board:
        return "draw"
    return None

def moves():
    return [i for i,v in enumerate(board) if v == EMPTY]

def minimax(maximizing):
    w = winner(board)
    if w == AI:    return 1
    if w == HUMAN: return -1
    if w == "draw": return 0
    
    if maximizing:
        best = -math.inf
        for m in moves():
            board[m] = AI
            best = max(best, minimax(False))
            board[m] = EMPTY
        return best
    else:
        best = math.inf
        for m in moves():
            board[m] = HUMAN
            best = min(best, minimax(True))
            board[m] = EMPTY
        return best

def best_move():
    best_score = -math.inf
    move_chosen = None

    for m in moves():
        board[m] = AI
        score = minimax(False)
        board[m] = EMPTY
        if score > best_score:
            best_score = score
            move_chosen = m

    return move_chosen

def play():
    print("You are O, AI is X. Positions:")
    print("0 | 1 | 2\n3 | 4 | 5\n6 | 7 | 8\n")
    turn = HUMAN

    while True:
        print_board()
        if turn == HUMAN:
            m = int(input("Your move (0-8): "))
            if m not in moves():
                print("Invalid move.\n")
                continue
            board[m] = HUMAN
        else:
            m = best_move()
            print(f"AI plays at {m}\n")
            board[m] = AI
        
        w = winner(board)
        if w:
            print_board()
            if w == "draw":
                print("Game drawn.")
            else:
                print(f"{w} wins!")
            break
        
        turn = AI if turn == HUMAN else HUMAN

if __name__ == "__main__":
    play()7
    
```
