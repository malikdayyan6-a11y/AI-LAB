```python
import heapq

def ucs(graph, start, goal, costs=None):
    """
    Uniform Cost Search implementation
    """
    if costs is None:
        costs = {}
        for node in graph:
            for neighbor in graph[node]:
                costs[(node, neighbor)] = 1
            
    pq = [(0, start, [start])]
    visited = set()
    
    while pq:
        total_cost, node, path = heapq.heappop(pq)
        if node in visited:
            continue
        visited.add(node)
        print(f"visiting node {node}, Cost: {total_cost}, Path: {path}")
        
        if node == goal:
            print(f"Goal found! Total cost: {total_cost}, Path: {path}")
            return path, total_cost
            
        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                edge_cost = costs.get((node, neighbor), 1)
                new_cost = total_cost + edge_cost
                heapq.heappush(pq, (new_cost, neighbor, path + [neighbor]))
                
    print("Goal not found!")
    return None, float('inf')
graph = {
    0: [1, 2],
    1: [3, 4],
    2: [5, 6],
    3: [],
    4: [],
    5: [],
    6: []
}
costs = {
    (0, 1): 2,
    (0, 2): 1,
    (1, 3): 3,
    (1, 4): 1,
    (2, 5): 1,
    (2, 6): 1
}

print("== Uniform Cost Search ===")
print("Searching from 0 to 6 with custom costs:")
ucs_path, ucs_cost = ucs(graph, 0, 6, costs)
print(f"\nUCS found path with cost {ucs_cost}")
```

```python
def dfs(graph, start, goal, path=None, visited=None):
    """
    Depth-First Search implementation
    """
    if path is None:
        path = [start]
    if visited is None:
        visited = set()
    
    visited.add(start)
    print(f"visiting node {start}, Path: {path}")
    
    if start == goal:
        print(f"\nGoal '{goal}' found! Path: {path}")
        return path
    
    for neighbor in graph.get(start, []):
        if neighbor not in visited:
            new_path = path + [neighbor]
            result = dfs(graph, neighbor, goal, new_path, visited)
            if result is not None:
                return result
    
    return None

graph = {
    'F': ['A', 'B', 'C'],
    'A': ['D', 'E'],
    'B': ['H', 'I'],
    'C': ['J', 'G'],
    'D': [],
    'E': [],
    'H': [],
    'I': [],
    'J': [],
    'G': []
}

print("== Depth-First Search (DFS) ==")
print("Tree structure (from image):")
print("    F")
print("   / \\")
print("  A   B   C")
print(" / \\ / \\")
print("D  E H I J G")
print("\nSearching from 'F' to 'G':")
dfs_path = dfs(graph, 'F', 'G')
print("-" * 50)
```

```python
def dls(graph, start, goal, limit, path=None, depth=0):
    """
    Depth-Limited Search implementation
    """
    if path is None:
        path = [start]
    
    print(f"Depth {depth}: visiting {start}, Path: {path}")
    
    if start == goal:
        return path
    
    if depth >= limit:
        print(f"Reached depth limit {limit} at node {start}")
        return None
    
    for neighbor in graph.get(start, []):
        new_path = path + [neighbor]
        result = dls(graph, neighbor, goal, limit, new_path, depth + 1)
        if result is not None:
            return result
    
    return None

graph = {
    'F': ['A', 'B', 'C'],
    'A': ['D', 'E'],
    'B': ['H', 'I'],
    'C': ['J', 'G'],
    'D': [],
    'E': [],
    'H': [],
    'I': [],
    'J': [],
    'G': []
}

print("== Depth-Limited Search (DLS) ==")
print("Tree structure (from image):")
print("     F")
print("   / | \\")
print("  A  B  C")
print(" / | | /")
print("D E H I G")
print("Searching from 'F' to 'G':")
print("=" * 30)
print("Test 1: Depth limit = 1 (too shallow)")
result = dls(graph, 'F', 'G', 1)
if result:
    print(f"Goal found! Path: {result}")
else:
    print("X Goal 'G' not found within depth limit 1!")
    
print("=" * 30)
print("Test 2: Depth limit = 2 (exact depth to reach G)")
result = dls(graph, 'F', 'G', 2)
if result:
    print(f"\nGoal 'G' found at depth 2!")
    print(f"Path: {result}")
    print(f"Path length: {len(result)-1} edges")
else:
    print("Goal not found!")
print("=" * 30)
```

```python
def dls_for_iddfs(graph, start, goal, limit, depth=0, path=None):
    """
    DLS helper for IDDFS
    """
    if path is None:
        path = [start]
    
    if start == goal:
        return path
    
    if depth >= limit:
        return None
    
    for neighbor in graph.get(start, []):
        new_path = path + [neighbor]
        result = dls_for_iddfs(graph, neighbor, goal, limit, depth + 1, new_path)
        if result is not None:
            return result
    
    return None

def iddfs(graph, start, goal, max_depth=10):
    """
    Iterative Deepening DFS implementation
    """
    print(f"Starting Iterative Deepening DFS from '{start}' to '{goal}'")
    
    for depth in range(max_depth + 1):
        print(f"\nDepth limit = {depth}: ", end="")
        result = dls_for_iddfs(graph, start, goal, depth)
        if result is not None:
            print(f"Goal found!")
            print(f"Path: {result}")
            print(f"Found at depth: {depth}")
            print(f"Path length: {len(result)-1} edges")
            return result, depth
        else:
            print("Goal not found at this depth")
    
    print(f"\nGoal '{goal}' not found within max depth {max_depth}")
    return None, None

graph = {
    'F': ['A', 'B', 'C'],
    'A': ['D', 'E'],
    'B': ['H', 'I'],
    'C': ['J', 'G'],
    'D': [],
    'E': [],
    'H': [],
    'I': [],
    'J': [],
    'G': []
}

print("Iterative Deepening DFS (IDDFS)")
print("Tree structure (from image):")
print("    F")
print("   / \\")
print("  A   B   C")
print(" / \\ / \\")
print("D  E H I J G")
print("\nSearching from 'F' to 'G':")
iddfs(graph, 'F', 'G', 5)
```
