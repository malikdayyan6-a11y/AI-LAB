```python
import numpy as np
original_array = np.array([[0, 1], [2, 3]])
print("Original flattened array:", original_array)
max_value = np.max(original_array)
print("Maximum value of the above flattened array:", max_value)
min_value = np.min(original_array)
print("Minimum value of the above flattened array:", min_value)
```

```python
import numpy as np
M1 = np.array([[5, 5],
               [1, 2]])
M2 = np.array([[0, 1],
               [1, 0]])
result = np.dot(M1, M2)
transpose_result = result.T
print("Matrix M1:\n", M1)
print("\nMatrix M2:\n", M2)
print("\nResult of M1 x M2:\n", result)
print("\nTranspose of the resultant matrix:\n", transpose_result)
```

```python
import numpy as np
matrix = np.array([[6, 1, 1, 3],
                   [4, -2, 5, 1],
                   [2, 8, 7, 6],
                   [3, 1, 9, 7]])
inverse_matrix = np.linalg.inv(matrix)
print(inverse_matrix)
```
