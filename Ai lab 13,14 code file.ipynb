```python
# Import required libraries
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Create small dataset
# Input features: exercise hours & water intake
# Output: healthy (1) or not (0)
X = np.array([[2, 3], [1, 2], [3, 4], [0, 1], [4, 5]], dtype=float)
y = np.array([[1], [0], [1], [0], [1]], dtype=float)

# Build the ANN model
model = keras.Sequential([
    layers.Dense(4, activation='relu', input_shape=(2,)),  # Hidden layer with ReLU
    layers.Dense(1, activation='sigmoid')                   # Output layer with sigmoid
])

# Compile the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Train the model
model.fit(X, y, epochs=100, verbose=0)

# Make prediction for new data
new_data = np.array([[2.5, 3.5]])
prediction = model.predict(new_data)
print("Prediction (Health Status):", (prediction > 0.5).astype(int)[0][0])
```

```python
# Import required libraries
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Create small dataset
# Input features: income & credit score
# Output: eligible (1) or not (0)
X = np.array([[50000, 700], [30000, 600], [80000, 800], 
              [20000, 500], [60000, 750]], dtype=float)
y = np.array([[1], [0], [1], [0], [1]], dtype=float)

# Build the ANN model
model = keras.Sequential([
    layers.Dense(4, activation='relu', input_shape=(2,)),  # Hidden layer with ReLU
    layers.Dense(1, activation='sigmoid')                   # Output layer with sigmoid
])

# Compile the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Train the model
model.fit(X, y, epochs=100, verbose=0)

# Make prediction for new data
new_data = np.array([[55000, 720]])
prediction = model.predict(new_data)
print("Prediction (Loan Eligibility):", (prediction > 0.5).astype(int)[0][0])
```
